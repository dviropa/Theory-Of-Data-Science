```python
oil_df["Date"] = pd.to_datetime(oil_df["Date"]).dt.tz_localize(None)

oil_df.set_index("Date", inplace=True)

oil_df.head()
```

```python
# לוודא ששתי עמודות התאריך הן datetime
stocks_df["date"] = pd.to_datetime(
    stocks_df["date"],
    format="%d/%m/%Y"
)
stocks_df["date"] = pd.to_datetime(stocks_df["date"])
oil_df["Date"] = pd.to_datetime(oil_df["Date"])

# למצוא את טווח התאריכים של המניות
start_date = stocks_df["date"].min()
end_date = stocks_df["date"].max()

print(start_date, end_date)

# לחתוך את הנפט
oil_df = oil_df[
    (oil_df["Date"] >= start_date) &
    (oil_df["Date"] <= end_date)
].copy()
```
